```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName("AirLine Assignment-1").getOrCreate()
```

```python
sc=spark.sparkContext
```

```python
airlines=sc.textFile("/content/drive/MyDrive/CDAC ClassWork/Big Data/Dataset/airlines.csv")
```

```python
header=airlines.first()
print(header)
```

```python
rdd1air=airlines.filter(lambda a: a!=header)
rdd1air.take(15)
```

1. How can you calculate the averge revenue per seat for each year and quarter?

```python
year_by = rdd1air.map(lambda a: ((a.split(",")[0],a.split(",")[1]),float(a.split(",")[2])))
year_by.take(15)
```

```python
year_by.count()
```

```python
year_by_avg = year_by.reduceByKey(lambda a,b:a+b)
for i in year_by_avg.take(15):
  print(i)
```

```python
year_by_avg.count()
```

```python
year_by_avg=year_by_avg.sortBy(lambda a: a[0])
for i in year_by_avg.take(15):
  print(i)
```

2. What is the PySpark code to find the year and quarter with the highest average revenue per seat?

```python
highest_rev=year_by_avg.sortBy(lambda a: -a[1])
for i in highest_rev.take(15):
  print(i)
```

3. How can you calculate the total number of booked seats for each year and Quater?

```python
booked_seat=rdd1air.map(lambda a: ((a.split(',')[0],a.split(',')[1]),float(a.split(',')[3])))
booked_seat.take(10)
```

```python
total_booked_seats=booked_seat.reduceByKey(lambda a,a1 : a+a1)
for i in total_booked_seats.take(10):
  print(i)
```

4. What is the Pyspark code to determine the year and quarter with the highest total number of booked seats?


```python
total_booked_seats=total_booked_seats.sortBy(lambda a: -a[1])
for i in total_booked_seats.take(10):
  print(i)
```

5. How can you calculate the total revenue genrted foe each year and quarter(rev=average*total_Number_seats)?

```python
total_rev=rdd1air.map(lambda a:((a.split(',')[0],a.split(',')[1]),round(float(a.split(',')[2])*float(a.split(',')[3]),2)))
for i in total_rev.take(15):
  print(i)
```

6. What is the pyspark code to identify the year and quarter with the highest total revenue?

```python
highest_totalrev=total_rev.sortBy(lambda a: -a[1])
for i in highest_rev.take(10):
  print(i)
```

7. How can you find the average revenue per seat across different years using Pyspark?

```python
year_wise=rdd1air.map(lambda a:(a.split(',')[0],float(a.split(',')[2]) ))

highest_year=year_wise.reduceByKey(lambda a,b:round(a+b),2)
for i in highest_year.take(10):
  print(i)
```

8. What is the PySpark code to determine the year with the highest average revenue per seat?

```python
highest_year=highest_year.sortBy(lambda a: -a[1])
for i in highest_year.take(10):
  print(i)
```

9. How can you calculate the overall average revenue per seat for the entire dataset using PySpark?

```python
overall_avg=rdd1air.map(lambda a: float(a.split(',')[2]) )
overall_avgs=overall_avg.reduce(lambda a,b:a+b)
print(overall_avgs)
```

#PySQL

1. How can you create a database and table in PySQL to store the airline CSV data?

```python
airline_DF=spark.read.format("csv").option("header","True").option("inferschema","True").load("/content/drive/MyDrive/CDAC ClassWork/Big Data/Dataset/airlines.csv")
```

```python
airline_DF.printSchema()
```

```python
airline_DF.createOrReplaceTempView("airline_db")
```

2. What is the PySQL query to calculate the average revenue per seat for each year and quarter?

```python
avg_rev=spark.sql("select year,quarter,Avg_rev_per_seat from airline_db")
avg_rev.show(15)
```

3. How can you find the year and quarter with the highest average revenue per seat using PySQL?

```python
high_avg=spark.sql("select year,quarter,Avg_rev_per_seat from airline_db order by Avg_rev_per_seat desc")
high_avg.show(10)
```

4. What is the PySQL query to calculate the total number of booked seats for each year and quarter?

```python
total_num_book = spark.sql("select year,quarter,sum(booked_seats) as Total_seats from airline_db group by year,quarter order by year,quarter")
total_num_book.show(10)
```

5. How can you determine the year and quarter with the highest total number of booked seats using PySQL?

```python
highest_seats=spark.sql("select Year,Quarter,Booked_Seats from airline_db order by Booked_Seats desc")
highest_seats.show(5)
```

6. What is the PySQL query to calculate the total revenue generated for each year and quarter (revenue = average revenue per seat * total number of booked seats)?

```python
total_rev_year=spark.sql("Select Year,Quarter,round((Avg_rev_per_seat*booked_Seats)/1000000,2) as Revenue_M  from airline_db order by Year,Quarter")
total_rev_year.show(10)
```

7. How can you identify the year and quarter with the highest total revenue using PySQL?

```python
high_total_rev=spark.sql("Select Year,Quarter,round((Avg_rev_per_seat*booked_Seats)/1000000,2) as Revenue from airline_db order by Revenue desc")
high_total_rev.show(5)
```
