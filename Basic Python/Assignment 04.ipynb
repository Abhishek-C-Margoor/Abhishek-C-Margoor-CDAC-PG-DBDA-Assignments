1. Write a function to print all prime numbers between two numbers.

```python
def is_prime(a, b):
  list1=[]
  for i in range(a+1, b):
    count=0
    for j in range(2,i):
      if i%j==0:
        count=count+1
        break

    if count==0:
      list1.append(i)


  return list1

print(is_prime(1,20))
```

2. Write a program to check whether a number is a perfect number (sum of its divisors =
number).

```python
def is_perfect_number(num):
  if num<=1:
    return False
  sum_num=1
  for i in range(2,int(num**0.5)+1):
    if num%i==0:
      sum_num+=i
      if i*i!=num:
        sum_num+=num//i
  return sum_num == num

number=int(input("Enter the number: "))

if is_perfect_number(number):
  print("The number is perfect number")
else:
  print("The number is not a perfect number")
```

digits = number).

```python
def factorial(n):
  fact=0
  if n<=1:
    return 1
  else :
    fact=n*factorial(n-1)
  return fact

def check_strong(num):
  string= str(num)
  sum_num=0
  for i in string:
    digit=int(i)
    print(digit)
    sum_num=sum_num+factorial(digit)


  if sum_num == num:
    print(f"{num} is a Strong number")
  else:
    print(f"{num} is not a Strong number")

number1=int(input("Enter the number : "))
check_strong(number1)
```

5. Write a function to generate the first n prime numbers.

```python
def check_prime(n):
  primelist=[]
  num=2
  while len(primelist) < n:
    is_prime = True
    for i in range(2,int(num**0.5)+1):
      if num%i==0:
        is_prime=False
        break
    if is_prime:
        primelist.append(num)
    num+=1
  return primelist

a=int(input("Enter the Number till u want to check Prime number: " ))
print(f"The first {n} prime number are: {check_prime(a)}")
```

6. Write a program to find the LCM (Least Common Multiple) of two numbers.

```python
def check_LCM(a,b):
  maxb=max(a,b)
  i=maxb

  while i>=maxb:
    if i%a==0 and i%b==0:
      return i
      break
    else:
      i+=1

num,num1=int(input("Enter the number to check first LCM: ")),int(input("Enter the number to check second number: "))
print(f"The LCM of {num} and {num1} is: {check_LCM(num,num1)}")
```

7. Write a function that returns the HCF (GCD) of a list of numbers.

```python
def hcf(gcd_list):
  ans=0
  for i in range(2, min(gcd_list)+1):
    flag=True
    for j in gcd_list:
      if i%j!=0:
        flag=False
        break

      if flag==True:
        ans=i
    return ans

print(hcf([2,4,6,10]))
```

```python
import math

def hcf_list(numbers):
  if not numbers:
    return None

  result = numbers[0]
  for i in range(1, len(numbers)):
    result = math.gcd(result, numbers[i])
  return result

# Example usage:
my_list = [12, 18, 24, 30]
hcf_of_list = hcf_list(my_list)
print(f"The HCF of {my_list} is: {hcf_of_list}")

my_list2 = [7, 14, 21]
hcf_of_list2 = hcf_list(my_list2)
print(f"The HCF of {my_list2} is: {hcf_of_list2}")

my_list3 = [5, 10, 15, 20, 25]
hcf_of_list3 = hcf_list(my_list3)
print(f"The HCF of {my_list3} is: {hcf_of_list3}")
```

11. Write a function to remove all vowels from a string.

```python
def rem_vowels(strn):
  emptystrg=''
  vowels=['a','e','i','o','u','A','E','I','O','U']
  for i in strn:
    if i not in vowels:
      emptystrg+=i
  return emptystrg

string=input("Enter the String : ")
print("String after removing Vowels is")
rem_vowels(string)
```

12. Write a program to check whether two strings are anagrams of each other.

```python
def check_anagrams(a,b):
  a=a.lower()
  b=b.lower()
  if len(a) != len(b):
    print("The Strings are not anagrams")
  else:
    freq=[0]*26
    freq1=[0]*26
    for i in a:
      freq[ord(i)-ord('a')]+=1
    for j in b:
      freq1[ord(j)-ord('a')]+=1

    if freq==freq1:
      print("The Given Strings are Anagrams")
    else:
      print("The Given Strings are not Anagrams")

a,b=input("Enter the first string to check Anagram: "),input("Enter the Second string to check Anagram: ")

check_anagrams(a,b)
```

15. Write a function to count uppercase and lowercase letters in a string.

```python
def count_case(s):
  upper_count = 0
  lower_count = 0
  for char in s:
    if 'a' <= char <= 'z':
      lower_count += 1
    elif 'A' <= char <= 'Z':
      upper_count += 1
  print(f"The count of Uppercase is {upper_count}")
  print(f'The count of Lowercase is {lower_count}')

st=input("Enter the string: ")
count_case(st)
```

17. Write a program to print a pyramid of numbers.


```python
row = int(input("Enter the number of rows: "))
for i in range(1, row + 1):
    for j in range(row - i):
        print(" ", end="")
    for j in range(1, 2 * i):
        print("*", end="")
    print()
```

18. Write a function to print the first n rows of Pascal’s Triangle.


```python
def pascals_triangle(n):
  triangle=[]
  for i in range(n):
    row=[1]
    if triangle:
      last_row=triangle[-1]
      for j in range(len(last_row)-1):
        row.append(last_row[j]+last_row[j+1])
      row.append(1)
    triangle.append(row)
  for row in triangle:
    print(" ".join(map(str,row)).center(n * 3))

rows=int(input("Enter the number of Rows: "))
pascals_triangle(rows)
```

19. Write a function to generate a multiplication table of size n x n (nested loop).

```python
def multiplication_table(n):
  for i in range(1,n+1):
    for j in range(1,n+1):
      print(i*j, end='\t')
    print()

size=int(input("enter the size of the Multiplication table (n): "))
multiplication_table(size)
```
