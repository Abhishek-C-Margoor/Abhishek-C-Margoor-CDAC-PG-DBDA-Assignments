19. Write a function to count the number of vowels in a given string.

```python

def count_vowels(s):
  vowels=['a','e','i','o','u','A','E','I','O','U']
  h=0
  for i in s:
    for j in vowels:
      if j == i:
        h=h+1

  print(" the Vowels count is :", h)

  return h

string= input("Enter the string to count the vowels: ")
count_vowels(string)
```
