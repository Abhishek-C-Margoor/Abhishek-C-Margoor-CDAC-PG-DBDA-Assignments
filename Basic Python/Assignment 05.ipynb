1. Write a program to check whether a given number is a happy number.
(A happy number eventually reaches 1 when replaced by the sum of the squares of its digits
repeatedly.)

```python
def num_square(n):
  num=0
  while(n):
    digits = n % 10
    num = num + digits*digits
    n = n // 10
  return num

def isHappy(n):
  st=set()
  flag = 1
  while  n!=1 and n not in st:
    st.add(n)
    n =num_square(n)
    flag=0
  return st,flag

numb=int(input("Enter the number to check wheter its a Happy number: "))
lis,flag=isHappy(numb)
print("The list of number we get in the process of checking happy number:",lis)
if  flag== 1:
  print(f"The Given number{numb} is not a Happy number")
else:
    print(f"The Given number {numb} is a Happy number ")
```
