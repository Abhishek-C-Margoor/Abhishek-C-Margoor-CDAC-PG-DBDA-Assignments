```python
def check_prime(num):
    if num <= 1:
        return False
    elif num == 2 or num == 3:
        return True
    else:
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                return False
        return True   # Only return True after all checks

txt = int(input("Enter the number to check: "))

if check_prime(txt):
    print("The number is prime")
else:
    print("The number is not prime")
```

1. Prime Number Check
Write a function is_prime(n) that checks if a number is prime.

```python
def is_prime(n):
  if n<=1:
    print("The number is not prime")
  elif n==2 or n==3:
    print("The number is prime")
  else:
    for i in range(2,int(n**0.5)+1):
      if n % i==0:
        print("The number is not prime")
      else:
        print("The number is prime")
ni=int(input("Enter the number:"))
is_prime(ni)
```

2. Palindrome Number
Write a function to check whether a number is a palindrome (e.g., 121 → Palindrome).

```python
def check_palindrome(num):
  num1=num
  rev=0
  while num>0:
    i=num%10
    rev=rev*10+i
    num=num//10
  if rev==num1:
    print("The number is palindrome : {}".format(rev))
  else:
    print("The number is not a palindrome: {}".format(rev))

n=int(input("Enter the number to check if its palindrome "))
check_palindrome(n)
```

3. Palindrome String
Write a function to check if a string is a palindrome (&quot;madam&quot; → True).

```python
def check_palindrome_string(s):
  rev_string=s[::-1]
  if rev_string == s:
    print("The Given String is Palindrome: {}".format(rev_string))
    print("True")
  else:
    print("The Given String is not Palindrome: {}".format(rev_string))
    print("False")

strng=input("Enter the string to check if its palindrome ")
check_palindrome_string(strng)
```

4. Factorial of a Number
Write a function factorial(n) that calculates factorial using a loop.

```python
def factorial(n):
  if n == 0 or n == 1:
    return 1
  else:
    return n * factorial(n - 1)

num=int(input("Enter the number:"))
print("The factorial of {} is: {}".format(num,factorial(num)))
```

5. Fibonacci Series
Write a function to print the first n terms of the Fibonacci series.

```python
def fibonacci(num):
  a=0
  b=1
  list1=[]
  for _ in range(num):
    list1.append(a)
    a,b=b,b+a
  return list1

numb=int(input("Enter the number:"))
fibonacci(numb)
```

6. Armstrong Number
Write a function to check if a number is an Armstrong number (e.g., 153 = 1³+5³+3³).


```python
def armst(num):
  strn=str(num)
  num_dig=len(strn)
  sum_pow=0
  for i in strn:
    digit=int(i)
    sum_pow=sum_pow + digit ** num_dig
  return sum_pow == num


number=int(input("Enter the number:"))
print(armst(number))

if armst(number):
  print("The number is armstrong")
else:
  print("The number is not armstrong")
```

8. Reverse a Number
Write a function to reverse a number (e.g., 123 → 321).

9. Count Vowels in a String
Write a function to count the number of vowels in a given string.

```python
def count_Vowels(txt):
  vowels=["a","e","i","o","u","A","E","I","O","U"]
  count=0
  for i in txt:
    if i in vowels:
      count+=1
  return count

string=input("Enter the string:")
count_Vowels(string)
```

10. Find GCD of Two Numbers
Write a function gcd(a, b) that finds the greatest common divisor of two numbers using
a loop.

```python
def gcd(a,b):
  c=min(a,b)
  ans=0
  for i in range(1,c+1):
    if a%i==0 and b%i==0:
      ans=i
  return ans

print(gcd(4,12))
```
